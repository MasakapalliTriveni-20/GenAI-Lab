```python
from transformers import pipeline

# Load the pre-trained GPT-2 model
generator = pipeline(
    "text-generation",
    model="gpt2"
)

# News headline/topic
headline = "India launches a new satellite to improve weather forecasting"

# Generate news content
generated_text = generator(
    headline,
    max_length=150,
    num_return_sequences=1,
    temperature=0.8
)

# Print the generated article
print("Generated News Article:\n")
print(generated_text[0]["generated_text"])
```
