```python
print("======================================")
print("     COLLEGE ADMISSION CHATBOT")
print("======================================")
print("Ask about courses, eligibility, fees or admission.")
print("Type 'exit' to stop.")
print()

while True:

    question = input("Student: ").lower()

    if question == "exit":
        print("Chatbot: Thank you! All the best.")
        break

    if "course" in question or "branch" in question:
        print("Chatbot: We offer CSE, AI & ML, ECE, EEE, Mechanical and Civil Engineering.")

    elif "eligibility" in question or "eligible" in question:
        print("Chatbot: Students who have completed 12th standard with the required qualification are eligible.")

    elif "fee" in question or "fees" in question:
        print("Chatbot: The course fees depend on the branch and admission category.")

    elif "admission" in question or "apply" in question:
        print("Chatbot: Admission includes application, counselling, document verification and fee payment.")

    else:
        print("Chatbot: Please ask about courses, eligibility, fees or admission.")

    print()
```
