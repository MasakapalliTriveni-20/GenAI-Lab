```python
from transformers import pipeline

# Load sentiment analysis pipeline
classifier = pipeline(
    "sentiment-analysis"
)

# Sample texts
texts = [
    "I absolutely loved the movie. It was fantastic!",
    "The product is okay, but it could be better.",
    "I am very disappointed with the customer service.",
    "This is the best day of my life!",
    "The experience was terrible and I will never return."
]

# Analyze sentiments
for text in texts:
    result = classifier(text)

    print(f"Text: {text}")
    print(f"Sentiment: {result[0]['label']}")
    print(f"Score: {result[0]['score']:.4f}")
    print("-" * 50)
```
