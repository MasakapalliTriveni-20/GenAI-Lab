```python
print("======================================")
print("       MEDICAL SYMPTOM CHECKER")
print("======================================")
print("Enter your symptoms separated by comma.")
print("Example: fever, cough, headache")
print()

symptoms = input("Enter symptoms: ").lower().split(",")

symptoms = [s.strip() for s in symptoms]

print()
print("Symptoms:", ", ".join(symptoms))
print()

# Rule-based reasoning

if "fever" in symptoms and "cough" in symptoms:
    result = "Possible Flu or Respiratory Infection"

elif "fever" in symptoms and "headache" in symptoms:
    result = "Possible Viral Infection"

elif "cold" in symptoms and "fever" in symptoms:
    result = "Possible Common Cold or Flu"

elif "sneezing" in symptoms and "runny nose" in symptoms:
    result = "Possible Allergy"

elif "stomach pain" in symptoms and "vomiting" in symptoms:
    result = "Possible Stomach Infection"

elif "headache" in symptoms:
    result = "Headache may have several causes such as stress, dehydration or lack of sleep"

elif "cough" in symptoms:
    result = "Cough may be caused by a cold, allergy or respiratory infection"

elif "fever" in symptoms:
    result = "Fever may be associated with an infection"

elif "cold" in symptoms:
    result = "Cold symptoms may be caused by a viral infection or allergy"

elif "sore throat" in symptoms:
    result = "Sore throat may be caused by a cold, flu or throat infection"

elif "back pain" in symptoms:
    result = "Back pain can have several possible causes"

elif "stomach pain" in symptoms:
    result = "Stomach pain can have several possible causes"

elif "fatigue" in symptoms:
    result = "Fatigue can be caused by lack of sleep, stress or illness"

else:
    result = "This symptom is not in my knowledge base. Please consult a healthcare professional."

print("Suggested Condition:", result)
print()
print("Note: This is an educational rule-based system, not a medical diagnosis.")
```
