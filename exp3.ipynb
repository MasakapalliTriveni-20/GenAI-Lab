```python
import random
import matplotlib.pyplot as plt

temperature = 25
temperatures = []
ac_status = []

for i in range(20):

    if temperature > 24:
        ac = "ON"
        temperature -= random.uniform(1, 2)

    elif temperature < 22:
        ac = "OFF"
        temperature += random.uniform(0.5, 1)

    else:
        ac = "OFF"
        temperature += random.uniform(0, 0.5)

    temperatures.append(temperature)
    ac_status.append(ac)

    print("Temperature:", round(temperature, 2), "°C | AC:", ac)

plt.plot(temperatures, marker='o')
plt.axhline(24, linestyle='--')
plt.axhline(22, linestyle='--')
plt.xlabel("Time")
plt.ylabel("Temperature (°C)")
plt.title("Smart Home Temperature Control")
plt.show()
```
